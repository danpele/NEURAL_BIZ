```python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')
# course chart style: transparent background, no gridlines, legend at the bottom outside the axes, four-digit years
plt.rcParams.update({
    'figure.facecolor': 'none', 'axes.facecolor': 'none', 'savefig.transparent': True,
    'axes.grid': False, 'axes.spines.top': False, 'axes.spines.right': False,
    'axes.edgecolor': '#222222', 'axes.labelcolor': '#222222', 'xtick.color': '#222222', 'ytick.color': '#222222',
    'text.color': '#222222', 'font.size': 11, 'axes.labelsize': 11, 'legend.fontsize': 10.5, 'legend.frameon': False,
    'lines.linewidth': 1.4, 'font.family': 'DejaVu Sans'})
SERIES = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100', '#e87ba4', '#008300', '#6250d6', '#e34948']
PAL = {'data': '#2a78d6', 'model': '#eb6834', 'benchmark': '#1baf7a', 'alt': '#eda100',
       'train': '#2a78d6', 'valid': '#eb6834', 'test': '#6250d6', 'positive': '#e34948', 'negative': '#2a78d6'}


def legend_bottom(fig, ncol=4):
    h, l_ = [], []
    for ax in fig.axes:
        for hh, ll in zip(*ax.get_legend_handles_labels()):
            if ll not in l_:
                h.append(hh); l_.append(ll)
    if h:
        fig.legend(h, l_, loc='lower center', ncol=min(ncol, len(h)), bbox_to_anchor=(0.5, 0.0), frameon=False)


def show(key, label, value):
    # print a number of the slide; [key] = its name in lectures/Lk/Lk_numbers.json
    print(f'{label} = {value}   [{key}]')
```

Daily means of hourly day-ahead prices on the CET calendar (the doubled autumn hour is averaged).

```python
x = pd.read_csv(DATA + 'power_da_prices_hourly_utc.csv', parse_dates=['timestamp_utc']).set_index('timestamp_utc')
x.index = x.index.tz_convert('Europe/Brussels')
def day_matrix(s):
    df = pd.DataFrame({'v': s.values, 'd': s.index.tz_localize(None).normalize(), 'h': s.index.hour}, index=s.index)
    return df.groupby(['d', 'h'])['v'].mean().unstack('h').reindex(columns=range(24))
RO = day_matrix(x['RO']).loc[:'2026-10-31']
DE = day_matrix(x['DE_LU']).loc[:'2026-10-31']
dly, dde = RO.mean(axis=1), DE.mean(axis=1)
show('facts.mean_2019', 'mean price 2019 (EUR/MWh)', round(float(np.nanmean(RO.loc['2019'].values)), 2))
show('facts.mean_2022', 'mean price 2022 (EUR/MWh)', round(float(np.nanmean(RO.loc['2022'].values)), 2))
show('facts.max_daily', 'highest daily mean (EUR/MWh)', round(float(dly.max()), 2))
show('facts.corr_ro_de_daily', 'correlation RO-DE daily means', round(float(dly.corr(dde)), 3))
```

```python
fig, ax = plt.subplots(figsize=(6.4, 2.4))
ax.plot(dde.index, dde.values, color=PAL['benchmark'], lw=0.6, alpha=0.8, label='Germany-Luxembourg')
ax.plot(dly.index, dly.values, color=PAL['data'], lw=0.6, label='Romania')
ax.set_ylabel('Daily mean price (EUR/MWh)')
legend_bottom(fig)
fig.tight_layout(rect=(0, 0.1, 1, 1))
fig.savefig('price_history.png', dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```
