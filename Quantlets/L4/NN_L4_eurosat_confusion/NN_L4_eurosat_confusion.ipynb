Errors of the fine-tuned network: number of errors, largest confusion, macro-F1.

```python
import json, urllib.request
_F = 'eurosat_confusion_results.json'
R = json.load(open(_F)) if os.path.isfile(_F) else json.load(urllib.request.urlopen(REPO + 'Quantlets/L4/NN_L4_eurosat_confusion/' + _F))
NAMES = ['Annual crop', 'Forest', 'Herbaceous veg.', 'Highway', 'Industrial', 'Pasture', 'Permanent crop', 'Residential', 'River', 'Sea/lake']
cm = np.array(R['cm']); off = cm.copy(); np.fill_diagonal(off, 0)
show('eurosat.n_errors_ft', 'errors on the test tiles', int(off.sum()))
P = cm.diagonal() / cm.sum(0); Rc = cm.diagonal() / cm.sum(1)
show('eurosat.macro_f1_ft', 'macro-F1 (%)', round(100 * float(np.mean(2 * P * Rc / (P + Rc))), 2))
fig, ax = plt.subplots(figsize=(6.4, 3.3)); cmn = cm / cm.sum(1, keepdims=True) * 100
ax.imshow(cmn, cmap='Oranges', vmin=0, vmax=100, aspect='auto')
for i in range(10):
    for j in range(10):
        if cm[i, j] >= 1: ax.text(j, i, str(cm[i, j]), ha='center', va='center', fontsize=6.5, color='white' if cmn[i, j] > 50 else '#222222')
ax.set_xticks(range(10)); ax.set_yticks(range(10)); ax.set_xticklabels(NAMES, rotation=40, ha='right', fontsize=7.2); ax.set_yticklabels(NAMES, fontsize=7.2)
ax.set_xlabel('Predicted class', fontsize=9); ax.set_ylabel('True class', fontsize=9)
fig.savefig('eurosat_confusion.png', dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```
