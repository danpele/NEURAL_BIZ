Fashion-MNIST: the same validation split as the lecture (seed 20261108), two training images per class.

```python
VISION = REPO + 'data/raw/vision' if os.path.isdir(REPO + 'data') else 'vision'
os.makedirs(VISION, exist_ok=True)
import torch, torch.nn.functional as F, torchvision
fm = torchvision.datasets.FashionMNIST(VISION, train=True, download=True)
X = fm.data.numpy() / 255; y = fm.targets.numpy()
itr = np.random.RandomState(20261108).permutation(len(X))[:50000]
CLASSES = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
show('fmnist.n_train', 'training images', len(itr))
fig, axes = plt.subplots(2, 10, figsize=(6.4, 1.75))
for c in range(10):
    ids = np.where(y[itr] == c)[0][:2]
    for r in range(2):
        axes[r, c].imshow(X[itr][ids[r]], cmap='gray_r'); axes[r, c].set_xticks([]); axes[r, c].set_yticks([])
    axes[1, c].set_xlabel(CLASSES[c], fontsize=7.5)
fig.savefig('fmnist_samples.png', dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```
